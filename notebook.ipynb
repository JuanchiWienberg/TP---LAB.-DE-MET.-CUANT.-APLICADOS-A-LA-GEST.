**Dataset:** DataCo Smart Supply Chain

Esta sección inicial realiza la carga remota del dataset desde GitHub y ejecuta el primer inventario de estructura y calidad de datos.

```python
import numpy as np
import pandas as pd

# Configuración de visualización para que los decimales y montos se lean con claridad
pd.set_option("display.float_format", "{:,.2f}".format)
pd.set_option("display.max_columns", None)

# URL Raw del archivo subido al repositorio de GitHub
url_raw = "https://raw.githubusercontent.com/JuanchiWienberg/TP---LAB.-DE-MET.-CUANT.-APLICADOS-A-LA-GEST./main/DataCoSupplyChainDataset.csv"

# Carga del dataset indicando encoding latin-1 para evitar errores de caracteres especiales
df = pd.read_csv(url_raw, encoding="latin-1")

print("--- PRIMERAS 5 FILAS ---")
display(df.head())

print("--- ÚLTIMAS 5 FILAS ---")
display(df.tail())
```

### 1. Estructura y Dimensiones del Dataset
A continuación, analizamos las dimensiones generales de la tabla, los nombres de columnas y sus tipos de datos guardados.

```python
# Dimensiones de la tabla
print(f"Dimensiones del dataset: {df.shape[0]} filas y {df.shape[1]} columnas\n")

# Información general de variables y tipos
print("--- INFORMACIÓN DE COLUMNAS Y TIPOS DE DATOS ---")
df.info()

# Cobertura temporal (convirtiendo la fecha de pedido a datetime)
if "order date (DateOrders)" in df.columns:
    df["order date (DateOrders)"] = pd.to_datetime(
        df["order date (DateOrders)"]
    )
    print(
        f"\nPeríodo temporal cubierto: {df['order date (DateOrders)'].min()} a {df['order date (DateOrders)'].max()}"
    )

# Cobertura geográfica (países o regiones representadas)
if "Order Region" in df.columns:
    print(
        f"Regiones comerciales representadas ({df['Order Region'].nunique()} en total):"
    )
    print(df["Order Region"].unique())
```

### 2. Evaluacion de Calidad de Datos: Nulos, Duplicados y Descriptivos
Analizamos la presencia de valores faltantes y registros duplicados para dejar asentado el estado del dataset antes de las transformaciones.

```python
# Recuento de valores nulos por columna
nulos = df.isnull().sum()
print("--- COLUMNAS CON VALORES NULOS ---")
print(nulos[nulos > 0])

# Control de filas duplicadas
duplicados = df.duplicated().sum()
print(f"\nCantidad de filas totalmente duplicadas: {duplicados}")

# Resumen estadístico de variables numéricas
print("\n--- ESTADÍSTICAS DESCRIPTIVAS INICIALES ---")
display(df.describe().T)
```

### 3. Hallazgos Iniciales de Calidad
- **Completitud:** Se identificaron valores nulos en columnas específicas (como campos opcionales de dirección o estado de pedidos), los cuales deberán ser tratados en la etapa de limpieza.
- **Duplicados:** No se observan duplicados exactos en filas completas.
- **Formato:** Algunas fechas requirieron conversión explícita a tipo `datetime` para permitir su análisis temporal.
